```python
from openai import OpenAI

from dotenv import load_dotenv
import os
```

```python

load_dotenv('Keys.env')

api_key = os.getenv("OPENAI_API_KEY")
```

```python
documents = [
    "Python is used for data science",
    "Java is used for backend development",
    "React is used for frontend development",
    "Machine learning learns patterns from data",
    "Neaural networks are used for deep learning",
    "Deeplearning is a subset of machine learning",
]
```

```python
url = "https://api.openai.com/v1/embeddings"


headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {api_key}"
}


payload = {
    "model": "text-embedding-ada-002",
    "input": documents
}
```

OpenAI's embeddings endpoint officially accepts an array of strings as input, and data in the response is an array of embedding objects—one for each input.

```python
import requests


data = requests.post(
    url =url,
    headers = headers,
    json = payload
)
```

```python
print(data.status_code)
```

```python
if data.status_code == 200:
    result = data.json() # takes the API response stored in data and converts to JSON format
    print(result['data'])
else:
    print("Error occurred while fetching API data")
```

```python
Embeddings = {}

for doc, emb in zip(documents, result['data']):
    Embeddings[doc] = emb['embedding']

print(Embeddings)

print(len(Embeddings)) # 6 
```

```python
for doc, embedding in Embeddings.items():
    print(doc)
```

```python
query = "Which technology is used for machine learning? and tell me the subsets of Machine learning"
```

```python
payload_query = {
    "model" : "text-embedding-ada-002",
    "input" : query
}

query_search = requests.post(
    url = url,
    headers = headers,
    json = payload_query
)
```

```python
if query_search.status_code == 200:
    result_query = query_search.json()
    print(f'Embedding for query {query} is {query_search.json()["data"][0]["embedding"]}')
    query_embedding = query_search.json()['data'][0]['embedding']
    print(f'Length of query embedding is {len(query_embedding)}')
```

Semantic Search

```python
import numpy as np

def cosine_similarity(vec1, vec2):
    a = np.array(vec1)
    b = np.array(vec2)
    dot_product = np.dot(a,b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot_product / (norm_a * norm_b)
```

```python

for doc, embedding in Embeddings.items():
    similarity_score = cosine_similarity(query_embedding, embedding)
    print(f"Similarity Between")
    print(f"Query : {query}")
    print(f"Document : {doc}")
    print(f"Similarity Score : {similarity_score}")
    print("---------------------------------------------------")

```
