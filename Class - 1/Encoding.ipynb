### One-Hot Encoding


```python
from sklearn.preprocessing import OneHotEncoder

import numpy as np
```

```python
documents = [
    "the cat sat on the mat",
    "the dog sat on the mat",
    "the cat chased the dog",
    "the dog chased the cat"
]
```

```python
vocabulary = set( word for doc in documents for word in doc.lower().split() )
vocabulary = list(vocabulary)
vocabulary
```

```python
vocabulary_array = np.array(vocabulary).reshape(-1, 1) # encoder expects 2D arraay as input

encoder = OneHotEncoder(sparse_output=False) # for easy interpretation we use sparse_output = False
# else we need to convert the sparse matrix to dense using .toarray() method
vector_values = encoder.fit_transform(vocabulary_array)

print(vector_values) #print(vector_values.toarray()) if sparse_output=True
print(encoder.categories_)
```

```python
for doc in documents:
    for word in doc.lower().split():
        word_vector = encoder.transform([[word]])
        print(f"Word: {word}, Vector: {word_vector}")
```

```python
testing_document = ["the cat sat on the mat", "the dog chased the cat"]

testing_vocabulary = list(set(word for doc in testing_document for word in doc.lower().split()))

testing_vectors = encoder.transform(np.array(testing_vocabulary).reshape(-1, 1))
print(testing_vocabulary)
print(testing_vectors)
print(encoder.categories_)
```

#### checking out of vocabulary 

```python
testing_document_2 = ["the cat sat on the mat", "the dog chased the cat", "the rabbit jumped"]

testing_vocabulary_2 = list(set(word for doc in testing_document_2 for word in doc.lower().split()))
print(testing_vocabulary_2)
print(encoder.categories_)
```

```python
testing_vectors_2 = encoder.transform(np.array(testing_vocabulary_2).reshape(-1, 1)) 
print(testing_vectors_2)

# since few words are not present in the training vocabulary, they will be ignored and not encoded.
# we will get value error if we try to encode a word that is not present in the training vocabulary.
```

### Bag Of Words
##### CountVectorizer and TfidfVectorizer, because they are designed to work directly with a list of text documents.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
BOW_vectorizer = CountVectorizer()
BOW_document_vectors = BOW_vectorizer.fit_transform(documents)
print(f'original documents : {documents}')
print(f'BOW vectors : {BOW_document_vectors.toarray()}')
print(f'Feature names : {BOW_vectorizer.get_feature_names_out()}')
```

##### Out of Vocabulary : do not usually raise the same kind of “out-of-vocabulary” error for unknown words the way OneHotEncoder does,instead, they simply ignore words that were not seen during fit

```python
BOW_testing_document_2 = BOW_vectorizer.transform(testing_document_2)
print(f'testing documents : {testing_document_2}')
print(f'BOW vectors : {BOW_testing_document_2.toarray()}')
print(f'Feature names : {BOW_vectorizer.get_feature_names_out()}')
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer 

tfidf_vectorizer = TfidfVectorizer()

tifidf_vectors = tfidf_vectorizer.fit_transform(documents)
print(f'original documents : {documents}')
print(f'TF-IDF vectors : {tifidf_vectors.toarray()}')
print(f'Feature names : {tfidf_vectorizer.get_feature_names_out()}')
```

### Out of Vocabulary : 

```python
tfidf_testing_document_2 = tfidf_vectorizer.transform(testing_document_2)
print(f'testing documents : {testing_document_2}')
print(f'TF-IDF vectors : {tfidf_testing_document_2.toarray()}')
print(f'Feature names : {tfidf_vectorizer.get_feature_names_out()}')
```
