```python

from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

```

```python
documents = [
    "The cat sat on the mat.",
    "The dog barked loudly at night.",
    "The bird built a nest in the tree.",
    "A lion lives in the jungle.",
    "Elephants are the largest land animals.",
    "The teacher wrote notes on the board.",
    "Students completed their homework before class.",
    "The professor explained deep learning algorithms.",
    "Artificial intelligence powers many modern applications.",
    "Machine learning models require quality data.",
    "Python is easy to learn.",
    "SQL is used to query databases.",
    "Pandas simplifies data analysis tasks.",
    "NumPy provides efficient numerical operations.",
    "Data visualization helps identify trends.",
    "The software engineer deployed the application.",
    "The developer fixed the production bug.",
    "Cloud computing offers scalable infrastructure.",
    "Cybersecurity protects sensitive information.",
    "The server handled thousands of requests.",
    "The doctor examined the patient carefully.",
    "Healthy food improves overall well-being.",
    "Regular exercise keeps the body fit.",
    "The nurse checked the patient's temperature.",
    "The hospital installed new medical equipment.",
    "The chef prepared spicy Indian curry.",
    "Fresh vegetables are rich in nutrients.",
    "The customer ordered biryani for dinner.",
    "The restaurant served delicious desserts.",
    "Coffee tastes better when freshly brewed.",
    "The football team won the championship.",
    "Basketball players trained every morning.",
    "The athlete completed the marathon.",
    "Swimming improves cardiovascular fitness.",
    "The coach motivated the entire team.",
    "The company hired a new manager.",
    "Employees attended the weekly meeting.",
    "Sales increased during the holiday season.",
    "Marketing campaigns attracted new customers.",
    "The startup secured funding from investors.",
    "Heavy rain flooded several streets.",
    "The sun shines brightly in summer.",
    "Snow covered the mountains during winter.",
    "The wind moved the tree branches.",
    "People enjoyed the pleasant weather.",
    "The family traveled to Arizona for vacation.",
    "Children played in the playground.",
    "The movie received excellent reviews.",
    "Music makes people feel relaxed.",
    "Technology continues to evolve rapidly."
]
```

```python
tokenized_docs = [simple_preprocess(doc) for doc in documents] # to remove punctuation and convert to lowercase
tokenized_docs
```

```python
vocabulary = sorted(list(set( word for words in tokenized_docs for word in words))) 
vocabulary
```

```python
vocab_dict = {}
for  index, word in enumerate(vocabulary):
    vocab_dict[index] = word
vocab_dict
```

CBOW : Word2Vec learns from contexts, not from a list of words. The tokenized sentences provide those contexts.

```python
CBOW_model = Word2Vec(
                         sentences = tokenized_docs,
                         vector_size = 20,
                         window = 2,
                         min_count = 1,
                         workers = 4,
                         sg = 0,
                         epochs = 100
                            
)
```

```python
print(vocab_dict[0])  # Get the first word in the vocabulary
print(CBOW_model.wv[vocab_dict[0]])  # Get the vector representation of the first word in the vocabulary
print(CBOW_model.wv[vocab_dict[0]].shape)
```

 Most similar words

```python
print(CBOW_model.wv.most_similar(vocab_dict[0], topn=5))  # Get the top 5 most similar words to the first word in the vocabulary
```

According to the learned embeddings, the word "a" is closest to these words

```python
print(CBOW_model.wv.most_similar('developer'))
```

What does model.wv['cat'] return?

- It returns the learned embedding vector for the word "cat". The vector is a fixed-length NumPy array (for example, 20, 100, or 300 dimensions depending on vector_size) that captures the word's semantic relationships learned during training.

"What does most_similar() do?"

- It computes the cosine similarity between the embedding vector of the given word and all other word vectors, then returns the words with the highest similarity scores.

 Skip-Gram : Target --> Context
 

```python
Skip_gram_model = Word2Vec(
                            sentences = tokenized_docs,
                            vector_size = 20,
                            window= 2,
                            min_count = 1,
                            workers = 4,
                            sg = 1,
                            epochs = 100
)
```

```python
print(vocab_dict[12])
print(Skip_gram_model.wv[vocab_dict[12]])
print(Skip_gram_model.wv[vocab_dict[12]].shape)
```

```python
print(Skip_gram_model.wv.most_similar('developer'))
```

```python
print(Skip_gram_model.wv.most_similar(vocab_dict[12], topn=5))  # Get the top 5 most similar words to the 13th word in the vocabulary
```
