```python
from openai import OpenAI
```

```python
from dotenv import load_dotenv
import os
```

```python


load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
```

```python
Client = OpenAI(api_key=api_key)
```

```python
def ask_llm(prompt):
    response = Client.responses.create(
        model="gpt-5.6",
        input=prompt
    )
    return response.output_text
```

```python
response = ask_llm("Explain neural networks simply.")
print(response)
```

```python
prompt_anatomy = """
# ROLE

You are a customer-support classification assistant.

# TASK

Analyze the customer's message.

# CONTEXT

Our company sells industrial automation equipment.

# INSTRUCTIONS

Determine:
1. Customer sentiment
2. Main issue
3. Urgency
4. Recommended department

# CONSTRAINTS

Do not invent missing information.
Urgency must be Low, Medium, or High.

# OUTPUT FORMAT

Return JSON:

{
    "sentiment": "",
    "issue": "",
    "urgency": "",
    "department": ""
}

# CUSTOMER MESSAGE

"My machine is down because the valve you supplied
isn't actuating. Production has stopped."
"""
```

```python
response = ask_llm(prompt_anatomy)
print(response)
```

** Zero - Shot Prompting ** : Give the model a task withoput showing examples

```python
Zero_shot_prompt = """

You are a sentiment analyzer.

Analyze the customer's message.

'I love this product and would like to buy more, but shipping cost is little too high.'

Return :
Sentiment : Positive, Negative, Neutral
Feedback 
Changes the company need to make to increase sales

"""
```
